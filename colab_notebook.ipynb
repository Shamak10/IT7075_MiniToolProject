## Hosted model access using Colab Secrets (§3.5)
Add ANTHROPIC_API_KEY under the key (🔑) icon in the left sidebar -> Secrets, and enable notebook access for it, before running this cell.

```python
!pip install -q anthropic
```

```python
from google.colab import userdata
from anthropic import Anthropic

client = Anthropic(api_key=userdata.get('ANTHROPIC_API_KEY'))
msg = client.messages.create(
    model="claude-sonnet-4-5",
    max_tokens=100,
    messages=[{"role": "user", "content": "Say hello in one short sentence."}],
)
print(msg.content[0].text)
```
